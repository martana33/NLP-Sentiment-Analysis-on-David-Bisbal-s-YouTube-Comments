# Lexicon Sentiment Analysis
This script uses a lexicon-based sentiment analysis (originally entered at https://gist.github.com/1035399). To download the AFINN word list do:.

```python
filenameAFINN = 'AFINN-111_es.txt'
afinn = pd.read_csv(filenameAFINN,
                          header=None, # no column names
                          sep='\t',  # tab sepeated
                          names=['term', 'value']) #new column names
afinn = afinn.set_index('term')
afinn = afinn['value'].to_dict()
```

```python
type(afinn)
afinn
```

```python
pattern_split = re.compile(r"\W+")
```

```python
def sentiment(text):
    """
    Returns a float for sentiment strength based on the input text.
    Positive values are positive valence, negative value are negative valence.
    """
    words = pattern_split.split(text.lower())
    sentiments = [afinn.get(word, 0) for word in words]
    if sentiments:
        # How should you weight the individual word sentiments?
        # You could do N, sqrt(N) or 1 for example. Here I use sqrt(N)
        sentiment = float(sum(sentiments))/math.sqrt(len(sentiments))

    else:
        sentiment = 0
    return sentiment
```

Few examples by seeing if the dictionary is working correctly:

```python
text = "feliz"
sentiment(text)
```

```python
text = "Madrid es una ciudad muy bella y me gusta vivir aquí"
sentiment(text)
```

```python
text = "Me sentí muy mal ayer. Tuve mucho dolor de cabeza y hubo una gran tormenta"
sentiment(text)
```
